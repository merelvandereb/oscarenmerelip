# Theorie

Volgens [Newton's wet van afkoeling](https://nl.wikipedia.org/wiki/Afkoelingswet_van_Newton) is de snelheid waarmee een voorwerp afkoelt evenredig met het verschil in de temperatuur van het voorwerp ($T$) en de omgeving ($T_0$). We kunnen dit schrijven als:

$$
    \dot{Q} = -hA(T(t) - T_0),
$$

waarin 

- $\dot{Q}$ de warmtestroom in $\mathrm{W}$,
- $A$ het oppervlak waardoor koeling optreedt in $\mathrm{m}^2$,
- $h$ de warmteoverdrachtscoëfficiënt in $\mathrm{W/(m^2 K)}$.

Dit levert de differentiaalvergelijking

$$
    C\dot{T} = -hA(T(t) - T_0),
$$

met $C$ de warmtecapaciteit in $\mathrm{J/kg}$. Herschrijven met $\tau = \frac{C}{hA}$ levert:

$$
    -\tau\dot{T} = T(t) - T_0,
$$ (eq:dv1)

met als oplossing:

$$
    T(t) - T_0 = (T(0) - T_0)\text{e}^{-t/\tau}.
$$ (eq:dv1_opl)

We kunnen hieruit dus concluderen dat $\tau$ de karakteristieke tijdsduur is waarin de temperatuur van de buis een factor $\text{e}$ verlaagd ten opzichte van de omgevingstemperatuur. 

```{exercise} Oplossing DV controleren 
:label: ex-dvopl
1. Neem de afgeleide van \(T(t)\):

\[
T(t) = (T(0) - T_0)e^{-t/\tau} + T_0
\]

\[
\dot{T}(t) = (T(0) - T_0)\,\frac{d}{dt}\left(e^{-t/\tau}\right)
           = (T(0) - T_0)\left(-\frac{1}{\tau}\right)e^{-t/\tau}.
\]

Dus

\[
\dot{T}(t) = -\frac{1}{\tau}(T(0) - T_0)e^{-t/\tau}.
\]

2. Vul dit in in de differentiaalvergelijking:

\[
-\tau \dot{T}(t) = -\tau \left[-\frac{1}{\tau}(T(0) - T_0)e^{-t/\tau}\right] 
                 = (T(0) - T_0)e^{-t/\tau}.
\]

3. Maar dat is precies:

\[
T(t) - T_0 = (T(0) - T_0)e^{-t/\tau}.
\]

Daarmee is aangetoond dat vergelijking \eqref{eq:dv1_opl} inderdaad een oplossing is van \eqref{eq:dv1}.
```

We zijn hier voor het gemak uitgegaan van een $h$ die onafhankelijk is van de temperatuur. We weten echter dat warmtetransport door straling niet lineair gaat, maar als 

$$
    \dot{Q}_s = \epsilon \sigma A (T^4 - T_0^4).
$$ 

Voor kleine temperatuurverschillen ($\Delta T = T - T_0$) is dit te vereenvoudigen tot

$$
    \dot{Q}_s = \epsilon \sigma A ((T_0+\Delta T)^4 - T_0^4) \approx \epsilon \sigma 4A T_0^3 \Delta T.
$$

Zolang $\Delta T$ dus relatief klein is ten opzichte van $T_0$, kunnen we $h$ dus inderdaad als een constante beschouwen. 

```{exercise} Uitwerken
:label: ex_uitw

Laat zien dat bovenstaande geldt door eerst $(T_0+\Delta T)^4$ uit te schrijven, te substitueren en dan te bedenken dat $\Delta T$ klein is en $\Delta T^2$ dus nog kleiner.
```

```{solution} ex_uitw
\[
(T_0 + \Delta T)^4 
= T_0^4 
+ 4T_0^3 \Delta T 
+ 6T_0^2 (\Delta T)^2 
+ 4T_0 (\Delta T)^3 
+ (\Delta T)^4.
\]

Substitueer dit in de warmtestroom:

\[
\dot{Q}_s 
= \epsilon \sigma A \left[(T_0+\Delta T)^4 - T_0^4\right].
\]

Na substitutie valt \(T_0^4\) weg, zodat overblijft:

\[
\dot{Q}_s 
= \epsilon \sigma A \left(
4T_0^3 \Delta T
+ 6T_0^2 (\Delta T)^2
+ 4T_0 (\Delta T)^3
+ (\Delta T)^4
\right).
\]

Voor kleine temperatuurverschillen geldt dat \(\Delta T \ll T_0\).  
Daarom zijn termen als \((\Delta T)^2\), \((\Delta T)^3\) en \((\Delta T)^4\) veel kleiner dan de lineaire term en kunnen worden verwaarloosd.

We houden dan over:

\[
\dot{Q}_s \approx \epsilon \sigma A \, 4T_0^3 \Delta T.
\]

Dit is lineair in \(\Delta T\), waardoor de warmtestroom zich gedraagt alsof er een constante warmteoverdrachtscoëfficiënt \(h\) bestaat.
```

# Resultaten

`````python
# Hier de data en de analyse

import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit



def exp_func(t, A, tau):
    # vaste omgevingstemperatuur
    T_omg = 29.0  # bijvoorbeeld, kies ~laatste meetwaarde of gemeten lokaaltemperatuur
    return A * np.exp(-t/tau) + T_omg

tijd_1 = np.array([0,10,20,28,37,44,52,60,66,76,85,93,103,113,123,137,148,160, 
                  172,184,197,210,223,240,256,270,286,310,327,340,355,370,
                  404,426])
tijd_2 = np.array([47.1,45.5,45,44.4,43.2,42.7,41.2,40.7,40.2,39.7,39.2,38.7,
                   38.2,37.8,37.3,36.8,35.8,35.3,34.8,34.4,34.3,33.8,33.4,
                   32.8,32.4,31.9,31.4,31,30.6,30.3,30.0,29.5,29.3,28.9])

# goede startwaarden
A0 = tijd_2[0] - 29.0   # beginverschil
tau0 = 300.0            # orde van grootte in seconden

popt, pocv = curve_fit(exp_func, tijd_1, tijd_2, p0=[A0, tau0], bounds=([0, 0], [np.inf, np.inf]))
A_exp, tau_exp = popt

print("A_exp =", A_exp)
print("tau_exp =", tau_exp)

# plot
T_omg = 29.0
y_fit = exp_func(tijd_1, *popt)

plt.figure()
plt.xlabel('Tijd [s]')
plt.ylabel('Temperatuur [°C]')
plt.plot(tijd_1, tijd_2, 'o', label='datapunten')
plt.plot(tijd_1, y_fit, 'r-', 
         label=f'T = {A_exp:.2f} e^(-t/{tau_exp:.1f}) + {T_omg:.1f}')
plt.legend()
plt.show()


buitenoppervlak = 0.00974  # m^2, mantel + dop (zoals eerder uitgerekend)
warmtecapaciteit = 156     # J/K, b.v. voor massieve messingbuis

h_exp = warmtecapaciteit / (tau_exp * buitenoppervlak)
print("h_exp =", h_exp, "W/(m^2 K)")

`````

# Discussie en conclusie
Uit de afkoelcurve van de metalen buis is een tijdsconstante tau  bepaald waarmee, in combinatie met het oppervlak en de warmtecapaciteit van de buis, de warmteoverdrachtscoëfficiënt h is berekend. De uiteindelijke waarde ligt rond de 108 W/(m^2)K, wat aangeeft dat de buis relatief snel warmte verliest aan de omringende lucht. De afkoeling volgt goed het exponentiële gedrag dat voorspeld wordt door Newtons wet van afkoeling, wat bevestigt dat het model bruikbaar is voor dit temperatuurbereik.
De gevonden waarde van h is hoger dan verwacht op basis van natuurlijke convectie en straling alleen. Dit komt waarschijnlijk doordat de warmtecapaciteit C aanvankelijk is overschat: de buis is hol, maar in de berekening is eerst uitgegaan van een massieve cilinder. Een te grote C leidt direct tot een te grote h. Daarnaast kan het effectieve koeloppervlak kleiner zijn dan aangenomen, bijvoorbeeld door isolatie van de grijparm of doordat delen van de buis niet volledig vrij in de lucht hangen. Ook meetruis in de temperatuurdata en onzekerheid in de omgevingstemperatuur beïnvloeden de fit. Ondanks deze afwijkingen laat de proef duidelijk zien dat convectie het dominante warmtetransportmechanisme is, met straling als kleinere maar niet verwaarloosbare bijdrage.
